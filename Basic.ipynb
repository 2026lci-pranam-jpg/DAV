```python
import pandas as pd
```

```python
s=pd.Series([10,20,30,40,52])

print(len(s))

print("\nLocation :",s.iloc[0])

s = pd.Series(list(s) + [50])
print(s)

s = s.drop(s.index[-1])
print(s)

s = s.drop(1)
print(s)
```

```python
#2
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 40],
    "Salary": [50000, 60000, 70000, 80000],
    "Department": ["HR", "IT", "Finance", "IT"]
}
df=pd.DataFrame(data)
```

```python
df["Bonus"] = [5, 10, 15,20]
print(df)
```
