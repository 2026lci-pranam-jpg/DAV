```python
df['Shape of You'] = df['Shape of You'].fillna(df['Shape of You'].mean())
print(df)
```

```python
#mean
print("Mean Shape of You:", df['Shape of You'].median())
```

```python
print("Mean Shape of You:", df['Shape of You'].mean())
```

```python
#median
df['Shape of You'] = df['Shape of You'].fillna(df['Shape of You'].median())
print(df)
```

```python
df['Despacito'] = df['Despacito'].ffill()
print(df)
```

```python
max_val = df['Despacito'].max()
print(max_val)
```

```python
df_sample = df.head(10).copy()

# 2. Apply Label Encoder to the sample
encoder = LabelEncoder()
df_sample['Shape of You_Encoded'] = encoder.fit_transform(df_sample['Shape of You'])

# 3. Print the result
print(df_sample[['Shape of You', 'Shape of You_Encoded']])
print("Classes:", encoder.classes_)
```
